```python
import os
```

```python
folder = "results"
# delete all files in the folder that not contains 'gift-eval' 'artificial' 'utsd'
for subfolder in os.listdir(folder):
    for ctx_len in os.listdir(os.path.join(folder, subfolder)):
        if ctx_len==".DS_Store":
            continue
        for file in os.listdir(os.path.join(folder, subfolder, ctx_len)):
            if not any(x in file for x in ['gift_eval', 'artificial', 'utsd']):
                print(f"Deleting {file} in {subfolder}")
                os.remove(os.path.join(folder, subfolder, ctx_len, file))
```
